**Question 2.a**

After training, use the trained model to predict `gender` for the training observations using only `height`. What is the training accuracy?

````
# YOUR CODE HERE
# Encode gender as binary labels
label_encoder = LabelEncoder()
y_train = label_encoder.fit_transform(df_train["gender"])
y_test = label_encoder.transform(df_test["gender"])

# Select height as the only input feature
X_train = df_train[["height"]]
X_test = df_test[["height"]]

# Train the neural network
model = MLPClassifier(
    hidden_layer_sizes=(10,),
    activation="logistic",
    solver="lbfgs",
    random_state=42
)
model.fit(X_train, y_train)

# Calculate training accuracy
train_predictions = model.predict(X_train)
train_accuracy = accuracy_score(y_train, train_predictions)

print(f"Training accuracy: {train_accuracy:.4f}")
print(f"Training accuracy (%): {train_accuracy * 100:.2f}%")
````

Z-score is a normalization function. It is the value of a feature minus the average value for that feature (in the training set), divided by the standard deviation of that feature (in the training set). **Question 6.a**

### Question 7- Bonus (1 point) ###
Can you improve the test-set accuracy from Question 6 by at least **2 percentage points**?

Using the same input features and preprocessing as in Question 6, experiment with neural-network hyperparameters such as the number of hidden layers, the number of nodes in each hidden layer, the activation function, solver, or regularization strength. **Do not add additional input features**.

You may use a systematic tuning method such as `GridSearchCV` to compare different hyperparameter combinations. When selecting hyperparameters, use cross-validation on the training data rather than selecting a model based directly on test-set performance.

Report the test-set accuracy of your final model and write a short paragraph explaining:

- which hyperparameters you explored,
- how you selected the final model, and
- how its test accuracy compares with your result from Question 6.

If you do not reach the 2-percentage-point target, report the best result you obtained and explain your approach.




````
# YOUR CODE HERE
from sklearn.compose import ColumnTransformer
from sklearn.model_selection import StratifiedKFold

# Same features and preprocessing as Q6.
# Putting preprocessing inside the pipeline ensures it is fitted
# separately on each training fold during cross-validation.
preprocessor = ColumnTransformer([
    ("numeric", StandardScaler(), ["height", "watchtv"]),
    ("categorical", OneHotEncoder(
        sparse_output=False,
        handle_unknown="ignore"
    ), ["year", "eyecolor", "exercise"])
])

pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", MLPClassifier(random_state=42, max_iter=1000))
])

# Compare 36 combinations of model settings
param_grid = {
    "model__hidden_layer_sizes": [(10,), (20,), (20, 10)],
    "model__activation": ["logistic", "relu"],
    "model__solver": ["lbfgs", "adam"],
    "model__alpha": [0.0001, 0.01, 0.1]
}

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

search = GridSearchCV(
    estimator=pipeline,
    param_grid=param_grid,
    scoring="accuracy",
    cv=cv,
    n_jobs=-1
)

features_q7 = ["height", "watchtv", "year", "eyecolor", "exercise"]

# Choose settings using only training data.
# GridSearchCV then refits the best pipeline on the full training set.
search.fit(df_train[features_q7], y_train)

# Evaluate the selected model on the test set
predictions_q7 = search.predict(df_test[features_q7])
accuracy_q7 = accuracy_score(y_test, predictions_q7)

improvement_pp = (accuracy_q7 - accuracy_q6) * 100

best_settings = {
    key.replace("model__", ""): value
    for key, value in search.best_params_.items()
}

print("Best settings:", best_settings)
print(f"Best cross-validation accuracy: {search.best_score_ * 100:.2f}%")
print(f"Q6 test accuracy: {accuracy_q6 * 100:.2f}%")
print(f"Q7 test accuracy: {accuracy_q7 * 100:.2f}%")
print(f"Change from Q6: {improvement_pp:+.2f} percentage points")
print("Reached the 2-percentage-point target:", improvement_pp >= 2)
````

###ANSWER:
I used GridSearchCV with 5-fold cross-validation on the training data to compare 36 combinations of hidden-layer sizes, activation functions, solvers, and regularization strengths. I kept the same features and preprocessing as Q6. The model with the highest mean cross-validation accuracy, 84.65%, used one hidden layer with 10 nodes, ReLU activation, the Adam solver, and alpha = 0.1. Its test accuracy was 85.18%, compared with 82.41% in Q6. Using the unrounded accuracies, this represents an improvement of 2.76 percentage points, exceeding the required 2-percentage-point target.
